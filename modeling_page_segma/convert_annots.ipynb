```python
from pathlib import Path
for lbl in ['train', 'test', 'val']:
    labels_dir = Path(f"data/{lbl}/labels")

    for file in labels_dir.glob("*.txt"):
        lines = []

        with open(file) as f:
            for line in f:
                parts = line.strip().split()
                if parts[0] != "0":
                    parts[0] = "0"
                lines.append(" ".join(parts))

        with open(file, "w") as f:
            f.write("\n".join(lines))
```
